```python
today = pd.Timestamp.today()

dff['tenure_days'] = np.where(
    dff['cancellation_date'].notna(),
    
    (dff['cancellation_date'] - dff['subscription_start_date']).dt.days,
    
    (today - dff['subscription_start_date']).dt.days
)

avg_tenure = dff['tenure_days'].mean()
print("Avg Tenure (Days) = ", round(avg_tenure),0)
```

```python
# revenue at risk - revenu lost from the chured users

revenue_at_risk = dff.loc[dff['churn_flag']==1, 'monthly_charges'].sum()
print("Revenue At Risk (Rs 'K') =",revenue_at_risk)
```

```python
# avrage complaint per user

avg_complaints = dff['complaint_count'].sum() / dff['customerid'].nunique()
print("Avg Compliants Per User = ", round(avg_complaints, 2))
```

```python
# Encode escalations: Y = 1, everything else = 0
dff['escalations'] = np.where(
    dff['escalations'] == 'Y', 
    1, 
    0
)

# Keep only required columns and remove missing values
corr_dff = dff[['escalations', 'churn_flag']].dropna()

# Calculate correlation
correlation = corr_dff['escalations'].corr(corr_dff['churn_flag'])

print("Correlation between escalation vs churn is =", round(correlation, 2))
```

```python
# Churn Risk - create a column existing col
conditions = [
    (dff['churn_score'] < 50),
    (dff['churn_score'] >= 50) & (dff['churn_score'] < 70),
    (dff['churn_score'] >= 70)
]

choices = ['low', 'med', 'high']

dff['churn_risk'] = np.select(
    conditions,
    choices,
    default='unknown'
)
```
